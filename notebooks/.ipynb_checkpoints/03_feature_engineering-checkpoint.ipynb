```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\Anindita\OneDrive\Desktop\credit-risk-prediction\data\raw\application_train.csv")
```

```python
df.shape
```

```python
df.head()
```

```python
df[[
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE",
    "DAYS_BIRTH",
    "DAYS_EMPLOYED",
    "CNT_FAM_MEMBERS",
    "CNT_CHILDREN"
]].head()
```

# Feature 1: Age

```python
df['AGE_YEARS'] = -df['DAYS_BIRTH']/365
```

```python
df[['DAYS_BIRTH', 'AGE_YEARS']].head()
```

# Feature 2: Employment Years

```python
df["EMPLOYMENT_YEARS"] = -df["DAYS_EMPLOYED"] / 365
```

```python
df[["DAYS_EMPLOYED", "EMPLOYMENT_YEARS"]].head()
```

```python
(df["DAYS_EMPLOYED"] == 365243).sum()
```

```python
df["EMPLOYED_ANOMALY"] = (df["DAYS_EMPLOYED"] == 365243).astype(int)
```

```python
df.loc[df["DAYS_EMPLOYED"] == 365243, "DAYS_EMPLOYED"] = np.nan
```

```python
df["EMPLOYMENT_YEARS"] = -df["DAYS_EMPLOYED"] / 365
```

```python
df[[
    "DAYS_EMPLOYED",
    "EMPLOYMENT_YEARS",
    "EMPLOYED_ANOMALY"
]].head(10)
```

# Feature 3: Credit-to-Income Ratio

```python
df["CREDIT_INCOME_RATIO"] = (
    df["AMT_CREDIT"] / df["AMT_INCOME_TOTAL"]
)
```

```python
df[[
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "CREDIT_INCOME_RATIO"
]].head()
```

# Feature 4: Annuity-to-Income Ratio

```python
df["ANNUITY_INCOME_RATIO"] = (
    df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]
)
```

```python
df[[
    "AMT_INCOME_TOTAL",
    "AMT_ANNUITY",
    "ANNUITY_INCOME_RATIO"
]].head()
```

# Feature 5: Credit-to-Goods Ratio

```python
df["CREDIT_GOODS_RATIO"] = (
    df["AMT_CREDIT"] / df["AMT_GOODS_PRICE"]
)
```

```python
df[[
    "AMT_CREDIT",
    "AMT_GOODS_PRICE",
    "CREDIT_GOODS_RATIO"
]].head()
```

# Feature 6: Income per Family Member

```python
df["INCOME_PER_FAMILY_MEMBER"] = (
    df["AMT_INCOME_TOTAL"] / df["CNT_FAM_MEMBERS"]
)
```

```python
df[[
    "AMT_INCOME_TOTAL",
    "CNT_FAM_MEMBERS",
    "INCOME_PER_FAMILY_MEMBER"
]].head()
```

# Feature 7: Children Ratio

```python
df["CHILDREN_RATIO"] = (
    df["CNT_CHILDREN"] / df["CNT_FAM_MEMBERS"]
)
```

```python
df[[
    "CNT_CHILDREN",
    "CNT_FAM_MEMBERS",
    "CHILDREN_RATIO"
]].head()
```

```python
new_features = [
    "AGE_YEARS",
    "EMPLOYMENT_YEARS",
    "EMPLOYED_ANOMALY",
    "CREDIT_INCOME_RATIO",
    "ANNUITY_INCOME_RATIO",
    "CREDIT_GOODS_RATIO",
    "INCOME_PER_FAMILY_MEMBER",
    "CHILDREN_RATIO"
]

df[new_features].head()
```

```python
np.isinf(df[new_features].select_dtypes(include=np.number)).sum()
```

```python
df.replace([np.inf, -np.inf], np.nan, inplace=True)
```

```python
df[new_features].describe()
```

```python
df[new_features + ["TARGET"]].corr()["TARGET"].sort_values()
```
